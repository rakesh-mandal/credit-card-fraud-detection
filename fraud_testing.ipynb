```python
import joblib
import pandas as pd

# Load saved pipeline and threshold
artifact = joblib.load("fraud_model.joblib")
pipeline = artifact["pipeline"]
threshold = artifact["threshold"]

new_transaction = pd.DataFrame([{
    "Time": 50000,
    'V1': -2.31, 'V2': 1.95, 'V3': -1.61, 'V4': 3.99, 'V5': -0.52,
    'V6': -1.43, 'V7': -2.54, 'V8': 1.39, 'V9': -2.77, 'V10': -2.77,
    'V11': 3.20, 'V12': -2.90, 'V13': -0.60, 'V14': -4.29, 'V15': 0.39,
    'V16': -1.14, 'V17': -2.83, 'V18': -0.01, 'V19': 0.42, 'V20': 0.13,
    'V21': 0.52, 'V22': -0.04, 'V23': -0.47, 'V24': 0.32, 'V25': 0.04,
    'V26': 0.18, 'V27': 0.26, 'V28': -0.14,
    "Amount": 149.62
}])

# new_transaction IS ALREADY a DataFrame — use it directly
fraud_prob = pipeline.predict_proba(new_transaction)[:, 1][0]
is_fraud = fraud_prob >= threshold

print(f"Fraud Probability : {fraud_prob:.4f}")
print(f"Prediction        : {'FRAUD' if is_fraud else 'GENUINE'}")
```
